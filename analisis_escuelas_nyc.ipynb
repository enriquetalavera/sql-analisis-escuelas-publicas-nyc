Cada año, el alumnado de secundaria en Estados Unidos realiza los SAT, unas pruebas estandarizadas diseñadas para medir las competencias de lectura, matemáticas y escritura. Hay tres secciones —reading, math y writing—, cada una con una **puntuación máxima de 800 puntos**. Estas pruebas son muy importantes tanto para el estudiantado como para las universidades, ya que influyen de forma decisiva en los procesos de admisión.

Analizar el rendimiento de los centros educativos es clave para muchos perfiles: responsables de políticas y de educación, personal investigador, administraciones públicas e incluso familias que están valorando a qué escuela deberían asistir sus hijos.

Dispones de un conjunto de datos llamado `schools.csv`, cuyo contenido puedes ver a continuación.

Tu tarea es responder a tres preguntas clave sobre el rendimiento en los SAT de las escuelas públicas de la ciudad de Nueva York (NYC).

```python
# Re-run this cell 
import pandas as pd

# Read in the data
schools = pd.read_csv("schools.csv")

# Preview the data
schools.head()

# 1. Escuelas con >= 80% de 800 en matemáticas = >= 640 puntos
best_math_schools = schools[schools["average_math"] >= 640][["school_name", "average_math"]].sort_values("average_math", ascending=False)

# 2. Top 10 escuelas por total SAT = math + reading + writing
schools["total_SAT"] = schools["average_math"] + schools["average_reading"] + schools["average_writing"]
top_10_schools = schools[["school_name", "total_SAT"]].sort_values("total_SAT", ascending=False).head(10)

# 3. Borough con mayor desviación estándar del total_SAT
borough_stats = schools.groupby("borough")["total_SAT"].agg(
    num_schools = "count",
    average_SAT = "mean",
    std_SAT = "std"
).reset_index()

largest_std_dev = borough_stats.sort_values("std_SAT", ascending=False).head(1)

# Redondear a 2 decimales como pide
largest_std_dev["average_SAT"] = largest_std_dev["average_SAT"].round(2)
largest_std_dev["std_SAT"] = largest_std_dev["std_SAT"].round(2)
```
