# Analisis de Calificaciones - Escuelas Publicas de NYC

## Resumen
Analisis exploratorio de 375 escuelas publicas de NYC basado en resultados SAT (matematicas, lectura, escritura). Objetivo: identificar brechas de rendimiento por distrito y factores de variabilidad.

## Dataset
- Fuente: NYC Open Data (via DataCamp)
- Registros: 375 escuelas, 8 variables
- Variables clave: borough, school_name, average_math, average_reading, average_writing, percent_tested

## Proceso
1. Carga y limpieza con pandas
2. Calculo de KPIs: total_SAT, filtrado >=80% en matematicas
3. Ranking Top 10 y analisis de variabilidad por borough


```python
import pandas as pd

# Carga de datos
schools = pd.read_csv("data/schools.csv")
schools.head()
```

```python
# 1. Escuelas con >= 80% de 800 en matematicas (>=640 puntos)
best_math_schools = schools[schools["average_math"] >= 640][["school_name", "average_math"]].sort_values("average_math", ascending=False)
best_math_schools.head()
```

```python
# 2. Top 10 escuelas por SAT total
schools["total_SAT"] = schools["average_math"] + schools["average_reading"] + schools["average_writing"]
top_10_schools = schools[["school_name", "total_SAT"]].sort_values("total_SAT", ascending=False).head(10)
top_10_schools
```

```python
# 3. Borough con mayor desviacion estandar del total_SAT
borough_stats = schools.groupby("borough")["total_SAT"].agg(
    num_schools = "count",
    average_SAT = "mean",
    std_SAT = "std"
).reset_index()

largest_std_dev = borough_stats.sort_values("std_SAT", ascending=False).head(1)
largest_std_dev["average_SAT"] = largest_std_dev["average_SAT"].round(2)
largest_std_dev["std_SAT"] = largest_std_dev["std_SAT"].round(2)
largest_std_dev
```

## Hallazgos principales
- Top Math: Stuyvesant High School 754, Bronx High School of Science 714
- Top SAT Total: Stuyvesant 2144, Bronx Science 2041, Staten Island Tech 2041
- Mayor variabilidad: Manhattan con 89 escuelas, promedio 1340.13 y desv. 230.29
